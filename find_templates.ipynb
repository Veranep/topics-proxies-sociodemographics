```python
all_prompts = {}
for prompt_option in prompt_options:
    wild_prompts = set(df_wild.loc[df_wild[prompt_option]]["opening_prompt"].tolist())
    lmsys_prompts = set(
        df_lmsys.loc[df_lmsys[prompt_option]]["opening_prompt"].tolist()
    )
    share_prompts = set(
        df_share.loc[df_share[prompt_option]]["opening_prompt"].tolist()
    )
    prompts = list(share_prompts) + list(lmsys_prompts) + list(wild_prompts)
    all_prompts[prompt_option] = prompts
    print(
        prompt_option,
        len(wild_prompts),
        len(lmsys_prompts),
        len(share_prompts),
        len(wild_prompts) + len(lmsys_prompts) + len(share_prompts),
        len(prompts),
    )
```
